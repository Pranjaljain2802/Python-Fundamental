Comparision Operator

```python
a = int(input("Enter number"))
b = int(input("Enter Second number"))

print(f"First number is {a} \nsecond number is {b}")

print(a==b) # Equal to
print(a<b) # Less then
print(a>b) # Grater then
print(a<=b) # less or Equal to
print(a>=b) # Greater or Equal to
print(a!=b) # Not Equal to 
```

# Logical Operators
- and (when both the conditon are true)
- op (either one of them is true)

```python
age = int(input("Enter age"))
salary = int(input("Enter Salary"))

print(f"Age  {age} \nSalary {salary}")

print(age >= 18 and salary >= 30000)
print(age > 18 or salary >= 30000)
print(not(age>=18)) # Not change Ture to False or False or True


```
