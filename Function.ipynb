# Function

Block of code that performs a specific task

A function is a block of organized code written to carry out a specified task. A function may or may not return data.

Functions help break a program into smaller and modular chunks for better readability. 

Parameters are specified after the function name inside the parentheses.

We can add as many parameters as needed, separated by comma. 
In Python a function is defined using the def keyword


## Three types of functions in Python:-

1) Built-in function :- Python predefined functions that are readily available for use like min() , max() , sum() , print() etc.

2) User-Defined Functions:- Function that we define ourselves to perform a specific task. using def keyword

3) Anonymous functions : Function that is defined without a name. Anonymous functions are also called as lambda functions. They are not declared with the def keyword

# Anonymous functions 
- Function that is defined without a name. 
- Anonymous functions are also called as lambda functions. 
- They are not declared with the def keyword
- It can any number of arguments
- It will take only one expression 
- Result will automatically return

```python
# lambda arguments : expression
```

```python
add = lambda a:a+10
print(add(4))
```

```python
sum = lambda a,b:a+b
print(sum(4,5))
```

```python
# find the greated number

greatest = lambda a,b: a if a>b else b
print(greatest(4,2))
```

# Fuilter function

- Filter elements based on condition
- return elements where condition is true
- filter (function, iterables)

```python
num = [1,2,3,4,5]
res=filter(lambda x:x%2==0,num)
print(list(res))
```

```python
num = [1,42,13,4,5]
res=filter(lambda x:x>10,num)
print(list(res))
```

# Map funciton 

- It will apply the operation for every elemnts in itrable
- map(function,iterable)

```python
numbers = [1,2,3,4,5]
add = map(lambda x:x+x, numbers)
print(list(add))
```

```python
l1 = [1,2,3,4]
l2 = [4,3,2,1]
add = map(lambda x,y:x+y,l1,l2)
print(list(add))
```

```python
word = ["text","cat","dog"]
res=map(list,word)

print(list(res))
```

# Reduce()

- It apply operation cumulatively to every element of iterable and give single resutl
- need to import

```python
from functools import reduce

numbers = [1,2,3,4,5]
res= reduce(lambda a,b: a+b, numbers)
print(res)
```
