# Set{}

Proprties 

- Hetrotypes
- Mutable (Dynamic in size but no value assignment)
- Unordered
- No indexing and slicing
- No Dupication 

```python
s = {10,20,30,'true',True}
print(s)
print("Type of s is ",type(s))
```

```python
s
```

```python
s1={10,20,30,40,50}
print('Old set\n',s1)

s1.add(88)
print('Updated set\n',s1)

s1.add(88)
print('Duplicate value in not allow\n',s1)
```

```python
# update is an another method to add a value in set

s1.update([3,4,5,6]) # its is also itrables
print('Updated set\n',s1)
```

```python
s1.pop() # its remove any one value in a set beacuse its has no index 
```

```python
s1.remove(20) # removes the object, but the element is not present it will give error
print('Updated set\n',s1)
```

```python
s1.discard(5) # removes the object, but the element is not present it will not  give error
print('Updated set\n',s1)
```

Set Method

- Union
- Intersection
- Difference

```python
# Union -> return all the unique elemtns

A = {10,20,30,40}
B = {30,40,50,60}

print(A.union(B)) # by function 
print(A|B) # by Operator 
```

```python
# Intersection -> Common Elements

print(A.intersection(B)) # by function
print(A&B) # by Operator
```

```python
# differenct -> give all unique elements present on First set
print(A.difference(B)) # by function
print(A-B) # By Operator
```
