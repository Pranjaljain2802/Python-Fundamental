# Loop control statemnts

- break (Exits out of loop/terminates loop) 
- continue (skip that current iteration )


```python
for i in range(1,11):
    if i == 5:
        break
    print(i)
```

```python
for i in range(5):
    if i == 3:
        continue
    print(i)
```

```python
for i in range(1,11):
    if i == 3:
        continue
    if i == 8:
        break
    print(i)
```

# While Loop 

- Execute a block of code as long as the condtion is true

```python
# while condition:
    # statemnt
    
```

```python
i = 1

while i <=5:
    print(i)
    i = i+1
```

```python
i = 2

while i <=11:
    print(i)
    i = i + 2
    #  
```

```python
i = 2
total_sum = 0

while i <=11:
    print(i)
    total_sum+=i
    i+=2

print(f"total sum is : {total_sum}")
```

```python
# total and sum of digits

num = int(input("Enter the number "))

total = 0

while num > 0:
    digit = num % 10 
    total = digit + total
    num = num // 10   # Flore Division retuen round of value

print(total)
```

```python
# Product of digits

num = int(input("Enter the number "))

product = 1

while num > 0:
    digit = num % 10 
    product = digit * product
    num = num // 10   # Flore Division retuen round of value

print(product)
```

```python
# Palindrome ->  (121,madam,)

num = int(input("Enter number"))

orignal = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num//=10

if orignal == reverse:
    print(f"{orignal} is Palindrome")
else:
    print(f"{orignal} is not a Palindrome")

```

```python
# do by self for string


txt = input("Enter the string")
reverse = ''

for ch in txt:
    reverse = ch + reverse

if txt == reverse:
    print("Palendrome")
else:
    print("Not a Palendrone")
```

```python
# Factorial

num = int(input("Enter a number"))

fact = 1 
i = 1 
while i <= num:
    fact = fact * i
    i = i + 1
print(fact)
```

```python
# do by using for loop
num = int(input("Enter a number: "))

fact = 1

for i in range(1, num + 1):
    fact = fact * i

print(fact)
```
