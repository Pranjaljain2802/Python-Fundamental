## 1. Dictionaries (`dict`)
A dictionary is a mutable collection of key-value pairs where keys must be unique and immutable.

```python
# Defining a dictionary
std = {
    'name': 'rahul',
    'age': 25,
    'Marks': 89,
    'Course': 'Python',
    'location': 'Bangalore'
}
std
```

### `.get(key)`
- Safely accesses value without throwing a `KeyError` if key does not exist.
- Does **not** modify original dictionary.

```python
# Accessing non-existing key returns None without error
print(std.get('landmark'))
```

### `.setdefault(key, default)`
- Returns the value if key exists.
- Inserts key with default value (or `None`) if key does not exist.

```python
# Adds 'email' key with value
std.setdefault('email', 'xyz@gmail.com')
print(std)
```

```python
# Existing key returns current value without updating
print(std.setdefault('email'))
```

```python
# Key absent without default inserts None
std.setdefault('userid')
print(std)
```

### `.items()`
- Returns a view of `(key, value)` tuple pairs.

```python
std.items()
```

## 2. Conditional Statements (Decision Making)

Used to control flow of execution based on boolean conditions (`True` / `False`).

### 2.1 `if` Statement



```python
#if condition:
    # Code executed if condition is True

# else:
    # Code executed if condition is False


age = 15
if age >= 18:
    print("the person is eligible to vote")
```

### 2.2 `if-else` Statement

```python
# Voting Eligibility
name = input("Enter your name: ")
age = int(input("Age: "))

if age >= 18:
    print(f"{name} you are eligible to vote")
else:
    print("{name} sorry you are ineligible to vote")
```

```python
# Positive or Negative
num = int(input("Ente number to check its +ve or -ve"))
if num > 0:
    print("Given number is positive")
else:
    print("Given number is negative")
```

```python
# Even or Odd
num = int(input("Give number "))
if num % 2 == 0:
    print("Even Number")
else:
    print("Odd number")
```

```python
# Number comparison
a = 10
b = 20
if a > b:
    print("a is greater than b")
else:
    print("b is greater than a")
```

### 2.3 `if-elif-else` Statement (Multiple Conditions)

```python
# Positive, Zero, or Negative
num = 0
if num >= 1:
    print("Given number is positive")
elif num == 0:
    print("Given number is neither positive nor negative")
else:
    print('Given number is negative')
```

```python
# Grading System
marks = 82
if marks > 90 and marks <= 100:
    print("grade A")
elif marks > 75:
    print("grade B")
elif marks > 60:
    print('grade C')
elif marks > 35:
    print("grade D")
else:
    print("Failed")
```

```python
# Traffic Signal Controller
signal = input("Enter Signal: ").lower()
if signal == 'red':
    print("stop")
elif signal == 'yellow':
    print("wait")
elif signal == 'green':
    print("move")
else:
    print("invalid input")
```

```python
# Greatest of Three Numbers
a = 15
b = 27
c = 12

if a > b and a > c:
    print("a is greater")
elif b > a and b > c:
    print("b is greater")
else:
    print("c is greater")
```

### 2.4 Nested `if` Statements

```python
# Age & ID Verification
age = 21
has_id = True

if age > 18:
    if has_id:
        print("entry is allowed")
    else:
        print("entry is not allowed")
else:
    print("age is invalid")
```

```python
# ATM Transaction Simulation
pin = 1234
balance = 5000
withdrawl = 2000

if pin == 1234:
    print("pin is correct")
    if balance >= withdrawl:
        print("withdraw the amount")
    else:
        print("balance is less")
else:
    print("pin is incorrect")
```

## 3. Practical Applications

### 3.1 Employee Bonus Calculator

```python
salary = 65000.0
exp = 2

if salary < 0 or exp < 0:
    print("invalid input")
elif exp < 1:
    bonus = 0
elif exp < 3:
    bonus = salary * 0.5
elif exp < 6:
    bonus = salary * 0.10
else:
    bonus = salary * 0.15

if salary >= 0 and exp >= 0:
    print(f"Bonus: {bonus}")
    print(f"total salary: {bonus + salary:.2f}")
```

### 3.2 Calculator Program

```python
a = 20
b = 4
operator = "*"

if operator == "+":
    print("result:", a + b)
elif operator == "-":
    print("result:", a - b)
elif operator == "*":
    print("result:", a * b)
elif operator == "/":
    if b != 0:
        print("result:", a / b)
    else:
        print("invalid divisor (division by zero)")
else:
    print("invalid operator")
```
